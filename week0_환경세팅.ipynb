## 0-1. Jupyter를 처음 사용할 때 알아둘 것

- **셀(Cell)**은 글이나 코드를 담는 칸입니다. 코드 셀은 Shift + Enter로 실행합니다.
- **커널(Kernel)**은 코드 셀을 실제로 실행하는 Python 프로세스입니다.
- 위 셀에서 만든 변수는 같은 커널의 아래 셀에서 사용할 수 있으므로 위에서부터 순서대로 실행합니다.
- 커널을 바꾸거나 설정을 수정했다면 **Restart → Run All**로 처음부터 다시 실행합니다.
- 오류가 나오면 마지막 줄의 오류 이름과 메시지를 먼저 확인합니다.

노트북 오른쪽 위의 커널 이름은 뒤에서 만들 APPS LangGraph 또는 이 폴더의 .venv여야 합니다.

## 0-2. 가상환경과 커널 준비

설정 요소는 다음 순서로 연결됩니다.

~~~text
requirements.txt의 패키지 목록
        ↓ pip로 설치
프로젝트 전용 .venv
        ↓ Jupyter 커널로 등록하고 선택
이 노트북에서 import
~~~

가상환경은 다른 프로젝트와 패키지 버전이 충돌하지 않도록 수업용 패키지를 별도 폴더에 보관합니다. 아래 명령은 노트북 코드 셀이 아니라 VS Code의 **터미널**에서 처음 한 번만 실행합니다.

### Windows PowerShell

강의자료 폴더를 VS Code로 연 뒤 Terminal → New Terminal에서 실행합니다.

~~~powershell
py -3.12 -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install --upgrade pip
python -m pip install -r requirements.txt
python -m ipykernel install --user --name apps-langgraph --display-name "APPS LangGraph"
~~~

활성화 명령에서 실행 정책 오류가 나면 현재 터미널에만 다음 설정을 적용한 뒤 다시 활성화합니다.

~~~powershell
Set-ExecutionPolicy -Scope Process -ExecutionPolicy Bypass
.\.venv\Scripts\Activate.ps1
~~~

### macOS / Linux

강의자료 폴더를 VS Code로 연 뒤 Terminal → New Terminal에서 실행합니다.

~~~bash
python3 -m venv .venv
source .venv/bin/activate
python -m pip install --upgrade pip
python -m pip install -r requirements.txt
python -m ipykernel install --user --name apps-langgraph --display-name "APPS LangGraph"
~~~

터미널 입력줄 앞에 (.venv)가 보이면 활성화된 상태입니다.

### 커널 선택

패키지 설치가 끝나면 노트북 오른쪽 위의 **Select Kernel**을 눌러 다음 중 하나를 선택합니다.

- APPS LangGraph
- 현재 강의자료 폴더 안의 .venv

터미널에서 .venv를 활성화하는 것만으로 이미 열려 있는 노트북의 커널이 자동 변경되지는 않습니다. 커널 선택까지 완료한 뒤 이 노트북을 위에서부터 실행하세요.

## 0-3. 현재 Python과 커널 진단

다음 셀은 현재 커널이 사용하는 Python 버전과 실행 파일을 보여 주고, 이 폴더의 .venv를 사용 중인지 확인합니다. 설정이 다르더라도 노트북을 중단하지 않고 해결 방법을 출력합니다.

```python
# 현재 Python 버전과 커널 환경을 확인합니다.
import sys
from pathlib import Path

version = sys.version_info
current_environment = Path(sys.prefix).resolve()
project_environment = (Path.cwd() / ".venv").resolve()

print(f"Python 버전: {version.major}.{version.minor}.{version.micro}")
print("Python 실행 파일:", sys.executable)
print("현재 작업 폴더:", Path.cwd())
print("예상 가상환경:", project_environment)

SUPPORTED_PYTHON = (version.major, version.minor) in {(3, 11), (3, 12)}
USING_PROJECT_VENV = project_environment.exists() and current_environment == project_environment

if not SUPPORTED_PYTHON:
    print("⚠️ Python 3.11 또는 3.12 사용을 권장합니다.")
elif not project_environment.exists():
    print("⚠️ .venv가 없습니다. 위의 OS별 명령부터 실행하세요.")
elif not USING_PROJECT_VENV:
    print("⚠️ 다른 커널입니다. Select Kernel에서 APPS LangGraph를 선택하세요.")
else:
    print("✅ Python 버전과 .venv 커널을 확인했습니다.")
```

표시된 Python 실행 파일 경로 안에 .venv가 있고 마지막 줄이 ✅라면 올바른 커널을 사용 중입니다.

경고가 나오면 위의 OS별 명령을 다시 실행하기 전에 먼저 다음 두 가지를 확인하세요.

1. VS Code에서 강의자료 폴더 자체를 열었는가?
2. 노트북 오른쪽 위에서 APPS LangGraph 또는 .venv 커널을 선택했는가?

## 0-4. 핵심 패키지 버전 확인

패키지 설치는 앞의 터미널 단계에서 한 번만 수행합니다. 아래 셀은 자동 설치를 하지 않고 현재 커널에서 핵심 패키지를 찾을 수 있는지만 확인합니다.

```python
# 패키지를 설치하지 않고 현재 버전만 확인합니다.
import importlib.metadata as package_metadata

required_packages = [
    "langgraph",
    "langchain-core",
    "langchain-openai",
    "python-dotenv",
    "ipykernel",
]
missing_packages = []

for package in required_packages:
    try:
        version_text = package_metadata.version(package)
        print(f"✅ {package} {version_text}")
    except package_metadata.PackageNotFoundError:
        missing_packages.append(package)
        print(f"❌ {package}: 설치되지 않음")

PACKAGES_READY = not missing_packages
if not PACKAGES_READY:
    print("터미널에서 실행: python -m pip install -r requirements.txt")
```

모든 항목이 ✅라면 현재 커널에서 수업용 패키지를 사용할 수 있습니다.

❌ 항목이 있다면:

1. 현재 커널이 .venv인지 다시 확인합니다.
2. VS Code 터미널에서 .venv를 활성화합니다.
3. python -m pip install -r requirements.txt를 실행합니다.
4. 설치가 끝나면 노트북 커널을 Restart합니다.

이 노트북 안에서는 설치 명령을 다시 실행하지 않습니다.

## 0-5. OpenAI API 키 설정

API 키는 계정과 사용 권한을 확인하는 비밀 값입니다. 코드 셀에 직접 입력하지 말고 강의자료 폴더의 .env 파일에 보관합니다.

1. .env.example을 복사해 .env라는 이름으로 저장합니다.
2. placeholder를 본인의 키로 바꿉니다.

~~~dotenv
OPENAI_API_KEY=your_openai_api_key_here
~~~

보안 주의:

- .env와 API 키를 GitHub, 단톡방, 압축파일, 제출물에 포함하지 않습니다.
- 키 값을 print하거나 화면 캡처에 넣지 않습니다.
- 키가 노출됐다면 파일에서 지우는 것만으로는 부족합니다. 해당 서비스에서 키를 폐기하고 새로 발급합니다.
- 가능하면 교육용 Project를 사용하고 개인별 사용 한도를 설정합니다.

다음 셀은 .env가 있으면 읽고, 키 내용을 표시하지 않은 채 사용할 수 있는 키인지 여부만 확인합니다. 파일이 없거나 예시 placeholder가 그대로 있어도 오류를 발생시키지 않습니다.

```python
# .env를 읽되 키 내용은 출력하지 않습니다.
import os
from dotenv import load_dotenv

env_path = Path.cwd() / ".env"
if env_path.exists():
    load_dotenv(dotenv_path=env_path, override=True)
    print("✅ .env 파일을 읽었습니다.")
else:
    print("ℹ️ .env가 없습니다. 운영체제 환경변수도 확인합니다.")

openai_key = os.environ.get("OPENAI_API_KEY", "").strip()
placeholder_prefixes = ("your_", "본인의_")
looks_like_placeholder = openai_key.startswith(placeholder_prefixes)
HAS_OPENAI_KEY = bool(openai_key and not looks_like_placeholder)
del openai_key

if HAS_OPENAI_KEY:
    print("✅ OPENAI_API_KEY가 설정되어 있습니다. 값은 표시하지 않습니다.")
else:
    print("⏭️ 유효한 키가 없어 모델 연결 확인을 건너뜁니다.")
```

## 0-6. ChatOpenAI 객체와 연결 확인

ChatOpenAI(...)는 사용할 모델과 시간 제한 같은 **호출 설정을 준비**합니다. 이 단계에서는 모델 답변이 생성되지 않습니다.

실제 네트워크 요청은 llm.invoke(...)를 실행할 때 발생합니다. 이때 계정 정책에 따라 사용량이나 비용이 생길 수 있고, 인증·사용량 제한·네트워크 문제로 실패할 수도 있습니다.

아래 첫 번째 셀은 키가 있을 때만 모델 객체를 만듭니다. 두 번째 셀은 연결 확인을 위한 실제 호출 코드입니다. 키가 없으면 자동으로 건너뜁니다.

```python
# 모델 객체를 준비합니다. 아직 실제 요청은 보내지 않습니다.
from langchain_openai import ChatOpenAI

MODEL_NAME = "gpt-4o-mini"

if HAS_OPENAI_KEY:
    llm = ChatOpenAI(
        model=MODEL_NAME,
        temperature=0,
        timeout=30,
        max_retries=2,
    )
    print(f"✅ {MODEL_NAME} 호출 설정을 준비했습니다.")
else:
    llm = None
    print("⏭️ 키가 없어 ChatOpenAI 객체 생성을 건너뜁니다.")
```

```python
# 키가 있을 때만 실제 API를 한 번 호출해 연결을 확인합니다.
if not HAS_OPENAI_KEY:
    CONNECTION_CHECK_STATUS = "skipped_no_key"
    print("⏭️ OpenAI 연결 확인을 건너뜁니다.")
else:
    try:
        connection_reply = llm.invoke(
            "연결 확인입니다. '연결 성공'이라고 짧게 답하세요."
        )
        CONNECTION_CHECK_STATUS = "success"
        print("응답 타입:", type(connection_reply).__name__)
        print("응답 내용:", connection_reply.content)
        print("✅ OpenAI 연결 확인 완료")
    except Exception as error:
        CONNECTION_CHECK_STATUS = "failed"
        print("❌ OpenAI 호출 실패:", type(error).__name__)
        print("아래 오류 표와 README의 문제 해결 항목을 확인하세요.")
```

## 0-7. 자주 발생하는 오류

| 증상 | 먼저 확인할 것 |
|---|---|
| ModuleNotFoundError | 현재 커널이 .venv인지, requirements 설치가 끝났는지 확인 |
| .env 파일을 찾지 못함 | .env.example을 복사했는지, 강의자료 폴더를 열었는지 확인 |
| AuthenticationError 또는 HTTP 401 | 키 오타·공백·placeholder 여부 확인. 노출된 키는 폐기 후 재발급 |
| RateLimitError 또는 HTTP 429 | Project 사용량과 요청 한도를 확인한 뒤 재시도 |
| APITimeoutError 또는 ConnectionError | 인터넷, 학교망 방화벽, VPN 상태 확인 |
| 모델 이름 관련 오류 | 강사가 안내한 현재 수업용 모델 이름으로 MODEL_NAME 수정 |

질문할 때는 오류가 난 셀, 오류 메시지의 마지막 5줄, Python 버전, 선택한 커널 이름을 함께 전달합니다. API 키는 캡처하지 않습니다.
